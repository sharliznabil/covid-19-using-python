### Table of numeric values

```python
New_eg_data=df_eg.groupby("date")[['total_cases','new_cases','new_cases_smoothed','total_deaths','new_deaths','new_deaths_smoothed','total_tests','new_tests','total_vaccinations','population_density','median_age','aged_65_older','aged_70_older']].sum().reset_index()
New_eg_data
```

### plot for relations between the new cases and the total cases ****

```python
plt.scatter(New_eg_data["new_cases"],New_eg_data['total_cases'])
plt.title("The relationship between new_cases and total_cases  " ,fontsize=25,color='r')
plt.xlabel("new_cases",fontsize=20,color='r')
plt.ylabel("total_cases",fontsize=20,color='r')
plt.show()
```

```python
plt.figure(figsize=(10,7))
plt.plot(New_eg_data['total_cases'],New_eg_data["total_tests"],label="total_cases")
plt.title("The relationship between total_tests and total_cases " ,fontsize=30,color='c')
plt.xlabel("total_cases",fontsize=20,color='r')
plt.ylabel("total_tests",fontsize=20,color='r')
plt.show()
```
